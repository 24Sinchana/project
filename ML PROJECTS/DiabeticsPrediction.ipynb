```python
import numpy as np
```

```python
import pandas as pd  #creating the data frame
from sklearn.preprocessing import StandardScaler #standardize the data
from sklearn.model_selection import train_test_split #split the data into train data and test data
from sklearn import svm #support vector Machine
from sklearn.metrics import accuracy_score #evaluating our model
```

```python
dataset=pd.read_csv("diabetes.csv")
```

```python
dataset.head()
```

```python
#number of rows and columns in this datasets
dataset.shape
```

```python
#getting the statistical measures of the data
dataset.describe()
```

```python
#it checks how many examples are there for the 1(diabetics) outcome and 0 ouctcome(non diabetics)
dataset['Outcome'].value_counts()
```

```python
dataset.groupby('Outcome').mean()  #group the data based on 0 and 1 
```

```python
#seperating data and labels,if specifing the drop function you have to specify the axis.
X=dataset.drop(columns ='Outcome',axis=1)
Y=dataset['Outcome']
print(X)
```

```python
print(Y)
```

```python
#standardization the data to have to common range between them
```

```python
scaler=StandardScaler()
```

```python
scaler.fit(X)
```

```python
standardized_data = scaler.transform(X)
```

```python
print(standardized_data)
```

```python
X=standardized_data    #represents the data
Y=dataset['Outcome']   #represents the model
```

```python
print(X)
print(Y)
```

```python
#Train Test Split
```

```python
#0.2 reprsents the 20%of the data from the test data
#Y has 1 or 0
#random state--ypu want to split the data(index or serial number for splitting of the data)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size = 0.2,stratify=Y,random_state=2)
#here 20% data is test_data
#and the remaining 80% will be used for training.
#same split of data is obtained every time the code is executed&allows you to compare results consistently. 
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
#768-original data set, 614-- training data and 154--test data
```

```python
classifier=svm.SVC(kernel ='linear')
```

```python
classifier.fit(X_train,Y_train)
```

```python
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
print('Acuracy score of the training data:',training_data_accuracy)
```

```python
#out of 100 prediction model is predicting 79 times correctly
```

```python
#predict the unknown data
```

```python
#accuracy score on the test data
X_test_prediction = classifier.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,Y_test)
print('Acuracy score of the training data:',test_data_accuracy)
```
