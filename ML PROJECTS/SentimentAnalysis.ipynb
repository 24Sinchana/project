```python
#access the tokensizer and model seperately
import tensorflow as tf
from transformers import DistilBertTokenizer,TFAutoModelForSequenceClassification
#class used to load the model
#Tf prefix of this class will load the tensorflow model and if you use the AutoModelForSequenceClassification without Tf then the model will be loaded from Pytorch
```

```python
Tokenizer=DistilBertTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
```

```python
Tokenizer
```

```python
Model=TFAutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
Model
```

```python
inputs=Tokenizer("The sun is shinning and the birds are chripping in the trees -What a Beautiful Day!",return_tensors="tf")
inputs
```

```python
#there are 23 Tokens in the input and we have array of integers representing the Tokens
#Tokenizer also outputs the Attention mask indicates what part of sentences should be considered or attended to
```

```python
logits=Model(inputs).logits
```

```python
logits
```

```python
predicted_class =int(tf.math.argmax(logits,axis=1)[0])
```

```python
Model.config.id2label[predicted_class]
```
