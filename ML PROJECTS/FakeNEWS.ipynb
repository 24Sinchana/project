```python
import numpy as np
import pandas as pd
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from nltk.stem.porter import PorterStemmer
#1-fake 0-real
```

```python
import nltk
nltk.download('stopwords')
print(stopwords.words('english'))
```

```python
#data preprocessing
news_dataset=pd.read_csv('train.csv')
```

```python
news_dataset.shape
```

```python
#print 5 rows of the dataframe
news_dataset.head()
```

```python
#count the missing values
news_dataset.isnull().sum()
```

```python
# replacing the null values with empty string
news_dataset = news_dataset.fillna('')
```

```python
news_dataset['content'] = news_dataset['author']+' '+news_dataset['title']
```

```python
print(news_dataset['content'])
```

```python
# separating the data & label
X = news_dataset.drop(columns='label', axis=1)
Y = news_dataset['label']
```

```python
print(X)
print(Y)
```

```python
port_stem = PorterStemmer()
```

```python
def stemming(content):
    stemmed_content = re.sub('[^a-zA-Z]',' ',content)#remove numbers,quotes special symbols
    stemmed_content = stemmed_content.lower()#convert all the words into lowercase
    stemmed_content = stemmed_content.split()#convert to list
    stemmed_content = [port_stem.stem(word) for word in stemmed_content if not word in stopwords.words('english')]
    stemmed_content = ' '.join(stemmed_content) #join all the words
   
    return stemmed_content
     #take out all the words which are not in the stopwords
    #only the significant words will be retured to the port_stem function & will be reduced to the keywords
```

```python
#stemming to reducte the word into root word
news_dataset['content'] = news_dataset['content'].apply(stemming)
```

```python
print(news_dataset['content'])
```

```python
#separating the data and label
X = news_dataset['content'].values
Y = news_dataset['label'].values
```

```python
print(X)
```

```python
print(Y)
```

```python
Y.shape
```

```python
#converting textual data into numerical data
vectorizer=TfidfVectorizer() #Tf-term frequency idf-inverse documentery frequency(it reduces the value repeated words/values )
vectorizer.fit(X)
X=vectorizer.transform(X)
#Tf-term frequency-finds the number of repeating significat wards and assigns the value
```

```python
print(X)
```

```python
#Splitting the train and test data
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
#stratify is used to divide the fake news and real news in the same proportion
#random_state=split the data(reproduce particular value)
```

```python
#Training the Model:LogisticRegression model
model=LogisticRegression()
model.fit(X_train,Y_train)
```

```python
#evaluation for accuracy score of training data
X_train_prediction=model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
print('Accuracy score of training data:',training_data_accuracy)
```

```python
#evaluation for accuracy score of test data
X_test_prediction=model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,Y_test)
print('Accuracy score of test data:',test_data_accuracy)
```

```python
#Making the predict system
X_new = X_test[3]
prediction=model.predict(X_new)
print(prediction)
if (prediction[0]==0):
    print("The news is real")
else:
    print("The news is fake")
```
