```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt #data visulization libraries
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
customer_dataset = pd.read_csv('Mall_Customers.csv')
```

```python
#choosing the annual income and spending score column
X= customer_dataset.iloc[:,[3,4]].values
```

```python
print(X)
```

```python
#choosing number of clusters
#finding WCSS values for clusters
wcss=[] # within clusters sum of square(distance between each data points and centroid of clusters)
for i in range(1,11): #a ll values from 1 to 10
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(X) #fit the data
    wcss.append(kmeans.inertia_) #find the wss value for each cluters
```

```python
#plot an elbow graph
sns.set()
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Point')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
#sharp point is cluter point
#number of clusters is 5
#K=5


```

Training the K-Means Clustering model


```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=0)
#return label each data point based on their K++
Y = kmeans.fit_predict(X)
print(Y)
```

```python
#5 clusters->>0,1,2,3,4

#K-means will randomly plot clusters
#5 centroids
```

Visualization of clusters

```python
#plotting all the clusters
plt.figure(figsize=(8,8)) #X->Annual Column,Y->Spending Score
plt.scatter(X[Y==0,0], X[Y==0,1], s=50,c='green' ,label ='Cluster 1')#cluster 1
plt.scatter(X[Y==1,0], X[Y==1,1], s=50,c='red' ,label ='Cluster 2')
plt.scatter(X[Y==2,0], X[Y==2,1], s=50,c='blue' ,label ='Cluster 3')
plt.scatter(X[Y==3,0], X[Y==3,1], s=50,c='orange' ,label ='Cluster 4')
plt.scatter(X[Y==4,0], X[Y==4,1], s=50,c='yellow' ,label ='Cluster 5')

#plot centroids  #0=x_axis_values of centroids and #1=y_axis_values of centroids
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1], s=100, c='black', label='Centroids')

plt.title('Customer Clusters')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.show()
```
